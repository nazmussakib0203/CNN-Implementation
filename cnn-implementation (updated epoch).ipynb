```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(device)
```

```python
transform = transforms.Compose([
    transforms.Resize((128, 128)),
    transforms.ToTensor()
])

dataset = datasets.ImageFolder(
    root="/kaggle/working/carambola_dataset/Original Dataset",
    transform=transform
)

train_size = int(0.8 * len(dataset))
test_size = len(dataset) - train_size

train_ds, test_ds = torch.utils.data.random_split(
    dataset,
    [train_size, test_size],
    generator=torch.Generator().manual_seed(42)
)

train_dl = DataLoader(train_ds, batch_size=64, shuffle=True)
test_dl = DataLoader(test_ds, batch_size=64, shuffle=False)

print("Train:", len(train_ds))
print("Test:", len(test_ds))
print("Total:", len(dataset))
print("Classes:", dataset.classes)
```

```python
# CNN Build
class CarambolaCNN(nn.Module):
    def __init__(self):
        super().__init__()

        # Convolutional Block
        self.conv1 = nn.Conv2d(3, 8, kernel_size=3, padding=1)
        self.conv2 = nn.Conv2d(8, 16, kernel_size=3, padding=1)
        self.pool = nn.MaxPool2d(2, 2)

        # ANN / Classification Layer
        input_neurons = 16 * 32 * 32
        self.cls = nn.Linear(input_neurons, 5)

    def forward(self, x):
        x = self.conv1(x)
        x = F.relu(x)
        x = self.pool(x)

        x = self.conv2(x)
        x = F.relu(x)
        x = self.pool(x)

        x = torch.flatten(x, 1)
        x = self.cls(x)

        return x

model = CarambolaCNN().to(device)
print(model)
```

```python
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

for ep in range(10,20):
    model.train()
    correct, total = 0, 0

    for img, label in train_dl:
        img, label = img.to(device), label.to(device)

        optimizer.zero_grad()
        output = model(img)
        loss = loss_fn(output, label)

        loss.backward()
        optimizer.step()

        correct += torch.sum(torch.argmax(output, dim=1) == label).item()
        total += label.size(0)

    print(f"Ep {ep+1}: Train Acc {correct/total}")
```

```python
model.eval()

correct, total = 0, 0
test_loss = 0.0

with torch.no_grad():
    for img, label in test_dl:
        img, label = img.to(device), label.to(device)

        output = model(img)
        loss = loss_fn(output, label)

        test_loss += loss.item() * label.size(0)

        predicted = torch.argmax(output, dim=1)
        correct += (predicted == label).sum().item()
        total += label.size(0)

test_accuracy = correct / total
test_loss = test_loss / total
 
print(f"Test Accuracy: {test_accuracy * 100:.2f}%")
print(f"Test Loss: {test_loss:.4f}")
```
